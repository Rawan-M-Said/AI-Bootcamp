##**Array Operations**

###**Find the dot product between the given arrays:**

```python
# Data:
 #[ 57  55   4  85 -88].[[-24 -27  28  25  47]]
 #[[19 91 36],[-74 -42  48],[19 -51 -83]]*[[-89 -10 -75],[-84 -25 -10],[-96   0  44]]

a = np.array([57, 55, 4, 85, -88])
b = np.array([-24, -27, 28, 25, 47])
c = np.array([[19, 91, 36],[-74, -42, 48],[19, -51, -83]])
d = np.array([[-89, -10, -75],[-84, -25, -10],[-96, 0, 44]])

result_1 = np.dot(a,b)
result_2 = np.dot(c,d)

print(result_1, '\n')
print(result_2)
```

##**Create a 4X4 random array with mean =1.5 and standerd deviation = 4,then re-shape it into 2X8 array**

```python
arr = np.random.normal(1.5, 4, (4,4))  #normal(loc=mean, scale=std_dev, size)
print(arr,'\n')

arr_reshaped = arr.reshape(2,8)
print(arr_reshaped)
```

### **Given this 1D array , add a new dimension , then flat it back into 1D**

```python
#[  79 -130   94   53   84 -175 -146 -105  169   81]
```

```python
array = np.array([79, -130, 94, 53, 84, -175, -146, -105, 169, 81])
array = array[:, np.newaxis]
print(array, '\n')

array = array.reshape(-1)
print(array)
```

## **Modifying columns.**

### **Change every value in coloumn:"HasCrCard" into True if it's 1 and False if it's 0.**

```python
data['HasCrCard'].replace(1, 'True', inplace=True)
data['HasCrCard'].replace(0, 'False', inplace=True)
data
```

### **Add a new coloumn to data named "Rich" ,it should be as follows :**
### **If the customer has more than 5k in his balance and active and has a credit card and the estimated salary is lower than <= the balance ,then he's rich.**
### **Else , he's not rich.**

```python
data['Rich'] = np.where((data['Balance']>5000) & (data['IsActiveMember']==1) & (data['HasCrCard']==1) &(data['EstimatedSalary']<=data['Balance']),'True','False')
data
```
