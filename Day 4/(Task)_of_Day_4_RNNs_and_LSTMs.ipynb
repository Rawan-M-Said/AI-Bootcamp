```python
import numpy as np
from keras.datasets import imdb
from keras.models import Sequential
from keras.layers import Dense, SimpleRNN, LSTM, Embedding
from keras.preprocessing import sequence
from sklearn.model_selection import train_test_split
```

## Sentiment Analysis

```python
# Load the IMDb dataset

########################################################
### DETERMINE YOUR DESIRED WORD NUMBER IN YOUR VOCAB ###
### DETERMINE YOUR MAX NUMBER OF WORDS PER REVIEW    ###
########################################################
max_features = 10000   # the top 10000 words (vocab number)
max_len = 500          # Cut off reviews after 500 words

(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=max_features)

# Padding -> to ensure each input is of the same length, it adds blanks (spaces).
x_train = sequence.pad_sequences(x_train, maxlen=max_len)
x_test = sequence.pad_sequences(x_test, maxlen=max_len)
```

```python
# Build the Simple RNN model

rnn_model = Sequential()
##################################################
### ADD THE RNN LAYERS (EMBEDDING, RNN, DENSE) ###
##################################################
rnn_model.add(Embedding(max_features, 128, input_length=max_len)) # 128 -> Dimensions (relations)
rnn_model.add(SimpleRNN(128))          # 128 -> num of instances of the rnn model
rnn_model.add(Dense(1, activation='sigmoid'))

rnn_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train the Simple RNN model

##############################################################################
### FIT THE MODEL WITH THE DESIRED EPOCHS, BATCH SIZE AND VALIDATION SPLIT ###
##############################################################################
rnn_model.fit(x_train, y_train, epochs=20, batch_size=64, validation_split=0.2)
```

```python
# Build the LSTM model

lstm_model = Sequential()
##################################################
### ADD THE RNN LAYERS (EMBEDDING, RNN, DENSE) ###
##################################################
lstm_model.add(Embedding(max_features, 128, input_length=max_len))
lstm_model.add(LSTM(128))
lstm_model.add(Dense(1, activation='sigmoid'))

lstm_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train the LSTM model

##############################################################################
### FIT THE MODEL WITH THE DESIRED EPOCHS, BATCH SIZE AND VALIDATION SPLIT ###
##############################################################################
lstm_model.fit(x_train, y_train, epochs=20, batch_size=64, validation_split=0.2)
```

```python
# Evaluate the Simple RNN model
rnn_loss, rnn_acc = rnn_model.evaluate(x_test, y_test)
print(f'Simple RNN Test Accuracy: {rnn_acc:.4f}')

# Evaluate the LSTM model
lstm_loss, lstm_acc = lstm_model.evaluate(x_test, y_test)
print(f'LSTM Test Accuracy: {lstm_acc:.4f}')
```

```python
# Function to preprocess the user input review
def preprocess_review(review, max_len=max_len, max_features=max_features):
    word_index = imdb.get_word_index()
    words = review.lower().split()
    tokens = [word_index.get(word, 1) for word in words]  # 1 is used for OOV words
    tokens = [1 if i >= max_features else i for i in tokens]  # Handle words not in the top 10,000
    return sequence.pad_sequences([tokens], maxlen=max_len)
```

```python
# Get a user review and preprocess it
user_review = input("Enter a movie review: ")
user_review_processed = preprocess_review(user_review)

# Predict with Simple RNN model
rnn_prediction = rnn_model.predict(user_review_processed)
rnn_sentiment = "Positive" if rnn_prediction >= 0.5 else "Negative"
print(f'Simple RNN Sentiment Prediction: {rnn_sentiment}')

# Predict with LSTM model
lstm_prediction = lstm_model.predict(user_review_processed)
lstm_sentiment = "Positive" if lstm_prediction >= 0.5 else "Negative"
print(f'LSTM Sentiment Prediction: {lstm_sentiment}')
```
