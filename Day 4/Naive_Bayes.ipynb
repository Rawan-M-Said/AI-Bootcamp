# Import necessary libraries

```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
import numpy as np
```

# Load and vectorize data

```python
train_data = fetch_20newsgroups(subset='train')
test_data = fetch_20newsgroups(subset='test')

def print_dataset_samples(data, n_samples=3):
    """
    Print a few samples from the dataset to understand its structure.

    Args:
    - data: The dataset object.
    - n_samples: The number of samples to print.
    """
    print("Displaying a few sample documents from the dataset:")
    print("-" * 80)
    for i in range(n_samples):
        print(f"Sample {i+1}")
        print("-" * 40)
        print(f"Category: {data.target_names[data.target[i]]}")
        print("Content:")
        print(data.data[i][:500])
        print("-" * 80)
        print()

# Print a few sample documents from the training dataset
print("Training Data Samples:")
print_dataset_samples(train_data, n_samples=3)

# Convert the text documents to a matrix of token counts (Bag of Words model)
vectorizer = CountVectorizer()
X_train = vectorizer.fit_transform(train_data.data)
X_test = vectorizer.transform(test_data.data)
y_train = train_data.target
y_test = test_data.target
```

# Train model

```python
model = MultinomialNB()
model.fit(X_train, y_train)
```

# Evaluate model

```python
y_train_pred = model.predict(X_train)
train_accuracy = accuracy_score(y_train, y_train_pred)
print(f'Naive Bayes Training Accuracy: {train_accuracy:.2f}')

y_test_pred = model.predict(X_test)
test_accuracy = accuracy_score(y_test, y_test_pred)
print(f'Naive Bayes Test Accuracy: {test_accuracy:.2f}')
```

# Calculate accuracy on the test data

```python
def print_predictions(samples, true_labels, predicted_labels, data):
    """
    Print a few samples of predictions alongside their true labels.

    Args:
    - samples: The indices of the samples to display.
    - true_labels: The true category indices.
    - predicted_labels: The predicted category indices.
    - data: The dataset object for reference to category names.
    """
    print("\nDisplaying predictions on test samples:")
    print("-" * 80)
    for idx in samples:
        print(f"Sample Index: {idx}")
        print(f"True Category   : {data.target_names[true_labels[idx]]}")
        print(f"Predicted Category: {data.target_names[predicted_labels[idx]]}")
        print("Content:")
        print(test_data.data[idx][:500])
        print("-" * 80)
        print()
```

```python
np.random.seed(42)
sample_indices = np.random.choice(len(y_test), 3, replace=False)
print_predictions(sample_indices, y_test, y_test_pred, test_data)
```
